```python
# Create a data list of student scores
students = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 92},
    {"name": "Charlie", "score": 78},
    {"name": "Diana", "score": 95}
]

# Calculate total and average
total_score = sum(student["score"] for student in students)
average_score = total_score / len(students)

# Print a formatted report
print("=== STUDENT PERFORMANCE REPORT ===")
for student in students:
    status = "Passed with Distinction!" if student["score"] >= 90 else "Passed"
    print(f"• {student['name']}: {student['score']} marks — {status}")

print("-" * 35)
print(f"Class Average Score: {average_score:.2f}")
```

```python
class Student:
    def __init__ (self,name,age):
        self.name=name
        self.age=age
    
student1=Student("ali",20)
student2=Student("ahmed",22)

print(student1.name, student1.age)
print(student2.name, student2.age)
    
```
